# 피보나치 수열 문제

```python
def solution1(n):
        consequence = 0
        list = [0,1]
        for i in range(n-2):
                list.append((list[len(list)-1]) + (list[len(list)-2]))

        return (list[len(list)-1]) + (list[len(list)-2])
```

# 문자열에서 최소값, 최대값 찾기

```python
def solution3(s):
        s_list = list(map(int, (s.split())))
        return str(min(s_list)) +" " + str(max(s_list))
```

```python
solution3("1 2 3 4")
```

```python
solution3("-1 -2 -3 -4")
```

```python
solution3("-1 -1")
```

# 이진 변환 (실패)
##### **이진법으로 변환하는 과정에서 오류**

```python
def solution4(s):
        s_delete = []
        for i in list(s):
                if i == "1":
                        s_delete.append(i)
        print(s_delete)     

        return str(int("1" * (len(s_delete) // 2)) - abs(len(s_delete) % 2) - 1)
```
