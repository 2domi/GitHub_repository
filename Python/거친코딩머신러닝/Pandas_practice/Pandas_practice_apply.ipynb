# 사용자 함수를 이용한 집계 (중요!!)

```python
import math
```

```python
def average_round(list):
    return round((sum(list))/(len(list)),2)
```

```python
abalone_df[["diameter", "height", "whole_weight"]].apply(lambda x : average_round(x), axis=1)
```

```python
def Q1(x,y,z):
    global answer
    if x + y + z >= 1 : 
        answer = True
        return(answer)
    else:
        answer = False
        return(answer)
    
abalone_df_copy = abalone_df.copy()
answer = abalone_df[["diameter","height","whole_weight"]] \
    .apply(lambda x: Q1(x.iloc[0],x.iloc[1],x.iloc[2]), axis=1)
abalone_df_copy["answer"] = answer
abalone_df_copy
```
