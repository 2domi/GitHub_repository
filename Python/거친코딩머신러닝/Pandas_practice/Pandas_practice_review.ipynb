# 조건 필터링

### *int,float*

```python
df[df["x^2"] >= 1000] # x^2가 1000보다 큰 것 출력
```

### *str*

```python
df1 = pd.DataFrame({"name":["이도","이연","동희","윤하"],"age":[13,13,47,44]})
df1
```

```python
df1.name == "이도"
```

```python
df1[df1.name == "이도"]
```

### *조건 여러개*

#### and function

```python
df[(df["x^2"] >= 1000) & (df["x^2"] <= 2000)] # 괄호로 조건을 묶기
```

#### or function

```python
df[(df["x^2"] <= 100) | (df["x^2"] >= 9000)] # 괄호로 조건을 묶기
```
