```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings

plt.rc('font', family='Malgun Gothic')
plt.rc('axes', unicode_minus=False)
sns.set_style("whitegrid")
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("Jungiyun_RockScissorsPaperData.csv").copy()
df
```

```python
all = pd.concat([df["previous"],df["now"]]).value_counts(dropna=False)
all
```

```python
sns.barplot(x=all.index,
                   y=all.values,
                   color="cornflowerblue",
                   width=0.8,
                   edgecolor="black",
                   linewidth=2,
                   alpha=0.7)
plt.xlabel("Jungiyun's RSP", fontsize=12)
plt.ylabel("count", fontsize=12)
plt.title("Counting RSP", fontsize=17)
plt.show()
```
