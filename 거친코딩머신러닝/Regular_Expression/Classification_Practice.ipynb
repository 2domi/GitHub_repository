```python
from sklearn import datasets
import numpy as np
import matplotlib.pyplot as plt
```

```python
iris = datasets.load_iris()
list(iris.keys())
```

# Logistic Regression (로지스틱 회귀)

```python
x = iris["data"][:,3:]
y = (iris["target"]==2).astype("int") # iris["target"]에서 2번 인덱스에 속하는 속성만 쓴다는 의미
x.shape, y.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
LOG = LogisticRegression()

LOG.fit(x,y)
```

```python
x_new = np.linspace(0,3,1000).reshape(-1,1)
x_new.shape
```

```python
plt.scatter(x,y,
                  color="blue")
plt.plot(x_new,LOG.predict_proba(x_new),
             color="red")
```
