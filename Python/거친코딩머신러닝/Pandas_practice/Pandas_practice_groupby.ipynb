# Groupby 연습

```python
import os
import pandas as pd
import numpy as np
```

```python
os.listdir("C:/Python_VSCode_Cursor/거친코딩머신러닝/")
```

```python
abalone_src = "C:/Python_VSCode_Cursor/거친코딩머신러닝/abalone.data"

abalone_df = pd.read_csv(abalone_src, header=None, sep=",",
                         names=['sex','length','diameter','height', 
                                'whole_weight','shucked_weight','viscera_weight', 
                                'shell_weight','rings'])

abalone_df
```

### 데이터가 비어있는지 확인

```python
abalone_df.isnull()
```

```python
abalone_df.isnull().sum() # True = 1, False = 0
```

```python
abalone_df.isnull().sum().sum()
```

### 기술통계 확인

```python
abalone_df.describe()
```

### groupby

```python
grouped = abalone_df["whole_weight"].groupby(abalone_df["sex"])
# 성별을 기준으로 전복의 무게를 집계
grouped.describe()
```

```python
abalone_df.groupby(abalone_df["sex"]).mean()
```

```python
abalone_df.groupby("sex").mean() # 위 코드와 같은 역할 수행
```

### 중복 데이터 삭제

```python
abalone_df.duplicated().sum() # 중복된 데이터가 존재하지 않음
```

```python
duplicate_abalone = abalone_df.iloc[[0]] # 중복된 조개 데이터 생성
duplicate_abalone
```

```python
duplicate_abalone_df = pd.concat([abalone_df,duplicate_abalone], axis=0)
duplicate_abalone_df
```

```python
duplicate_abalone_df.duplicated() 
```

```python
duplicate_abalone_df.duplicated().sum()
```

### 중복 데이터 삭제

```python
duplicate_abalone_df.drop_duplicates(inplace=True)
duplicate_abalone_df
```

```python
(duplicate_abalone_df == abalone_df).sum() # 중복을 없앤 중복 생성 데이터와 원래 데이터가 같은지 확인
```

# 빈 데이터 삭제 (Nan)

```python
NaN_abalone_df = abalone_df.copy()

NaN_abalone_df
```

```python
NaN_abalone_df.iloc[0,1] = np.nan
NaN_abalone_df
```

```python
NaN_abalone_df.isnull().sum().sum()
```

```python
NaN_abalone_df.mean(numeric_only=True)
```

```python
NaN_abalone_df.fillna(NaN_abalone_df.mean(numeric_only=True))
```
