```python
import pandas as pd
import os
```

```python
column1 = []
column2 = []

for i in range(1,101) :
    column1.append(i**2)
    column2.append(i**3)

print(column1)    
print(column2)    
```

```python
all_index = []

for i in range(1,101):
    all_index.append(i)
```

```python
df = pd.DataFrame({"x":all_index,"x^2":column1,"x^3":column2})
df
```

```python
df.to_csv("Example.csv", index=False)
```
