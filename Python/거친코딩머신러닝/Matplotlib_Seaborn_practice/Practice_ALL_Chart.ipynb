```python
import pandas as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import matplotlib as rc
import seaborn as sns
import numpy as np
import warnings

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = [10,8]
font_path = r"c:\USERS\SAMSUNG\APPDATA\LOCAL\MICROSOFT\WINDOWS\FONTS\FREESENTATION-4REGULAR.ttf"
fontprop = lambda size: fm.FontProperties(fname=font_path,size=size)
```

## 예제 데이터셋 불러오기

```python
diamonds = sns.load_dataset("diamonds")
diamonds
```

# Bar Chart

```python
sns.barplot(data=diamonds,
                    x="cut",
                    y="price",
                    hue="color",
                    palette="rainbow",
                    alpha=1,
                    edgecolor="black",
                    linewidth=2,
                    order=["Ideal", "Premium", "Very Good", "Good", "Fair"])
plt.title("자른 방식에 따른 색상별 다이아몬드 가격",fontproperties=fontprop(30))
plt.xlabel("Cut", fontproperties=fontprop(20))
plt.ylabel("Price", fontproperties=fontprop(20))
plt.show()
```

# Histogram

```python
sns.histplot(data=diamonds,
                    x="price",
                    color="deepskyblue",
                    bins=60,
                    hue="cut",
                    multiple="stack",
                    palette="rainbow",
                    edgecolor="black",
                    linewidth=1,
                    alpha=0.8)
plt.title("자른 방식에 따른 가격 분포도",fontproperties=fontprop(30))
plt.xlabel("Price", fontproperties=fontprop(20))
plt.ylabel("Count", fontproperties=fontprop(20))
plt.show()
```

# Pie Chart

```python
cut__price = diamonds.groupby("cut")["price"].mean()

ratio_cut__price = cut__price / sum(cut__price) * 100
ratio_cut__price
```

```python
plt.pie(ratio_cut__price,
            labels=["Ideal","Premium","Very Good","Good","Fair"],
            autopct='%.1f%%',
            startangle=90,
            counterclock=False,)
plt.title("자른방식당 개수 비율",fontproperties=fontprop(30))
plt.show()
```

# Scatter Chart

```python
sample_diamonds = diamonds.sample(frac=0.05,random_state=0)
sample_diamonds
```

```python
sns.scatterplot(data=sample_diamonds,
                          x="carat",
                          y="price",
                          alpha=0.8,
                          hue="cut",
                          palette="rainbow",
                          s=30,
                          marker="o",
                          )
plt.title("Carat vs Price by Cut", fontproperties=fontprop(30))
plt.xlabel("Carat", fontproperties=fontprop(15))
plt.ylabel("Price", fontproperties=fontprop(15))
plt.xticks(fontproperties=fontprop(12))
plt.yticks(fontproperties=fontprop(12))
plt.grid(alpha=0.3)
plt.legend(title="Cut", fontsize=12, title_fontsize=13)
plt.tight_layout()
plt.show()
```

# Line Chart

```python
flights = sns.load_dataset("flights")
flights
```

```python
sns.lineplot(data=flights,
                    x="year",
                    y="passengers",
                    hue="month",
                    palette="rainbow",
                    alpha=0.7,
                    linewidth=2)
plt.plot(list(range(1949,1961)),
             flights.groupby("year")["passengers"].mean().values,
             color="black",
             linewidth=3)
plt.title("연도별 비행기 승객수",fontproperties=fontprop(30))
plt.xlabel("Year", fontproperties=fontprop(20))
plt.ylabel("Passengers", fontproperties=fontprop(20))
plt.xticks(fontproperties=fontprop(12))
plt.yticks(fontproperties=fontprop(12))
plt.show()
```
